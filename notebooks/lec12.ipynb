## Initializing numpy arrays for scalars, vectors and matrices

First we import the numpy library. It's a Python idiom to shorten numpy as `np`.

```python
import numpy as np
```

Essentially, you just wrap `np.array()` around a list.

- **Scalar**: Use `np.array()` with a single value in a list.
    - Example: `a = np.array([5])`

- **Vector**: Use `np.array()` with a one-dimensional list.
    - Example: `v = np.array([1, 2, 3])`

- **Matrix**: Use `np.array()` with a list of lists, where each inner list represents a row.
    - Example: `m = np.array([[1, 2], [3, 4]])`

To extract elements, rows, or columns from a matrix:

- **Access an element:** Use `matrix[row, column]`.
    - Example: `m[1, 0]` gets the element in the second row, first column.

- **Access a row:** Use `matrix[row, :]`.
    - Example: `m[0, :]` gets the first row.

- **Access a column:** Use matrix[:, column].
    - Example: m[:, 1] gets the second column.

- **Access a submatrix:** Use slicing matrix[start_row:end_row, start_col:end_col].
    - Example: m[0:2, 1:3] gets rows 0-1 and columns 1-2.

- **Modify elements:** Assign values directly using indexing.
    - Example: m[1, 1] = 10 sets the element in the second row, second column to 10.

# Why Numpy and not just lists?

We're going to add the two vectors `x = [1,2,3]` and `y = [4,5,6]`, using a list and numpy.

> Add blockquote



```python
x_list = [1, 2, 3]
y_list = [4, 5, 6]

# Using lists
z_list = []
for i in range(len(x_list)):
    z_list.append(x_list[i] + y_list[i])
print("Vector addition using lists:", z_list)

# Using numpy
x_np = np.array(x_list)
y_np = np.array(y_list)
z_np = x_np + y_np
print("Vector addition using numpy:", z_np)
```
