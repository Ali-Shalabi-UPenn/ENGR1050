# Problem statement

Together, we are going to design a class called `ODE_solver` which we will use to solve ODEs in the homework. The class should take as inputs:

- An initial condition $t_0$ and $y_0$. Note that the solution may either be scalar-valued or vector-valued.
    - Example: Scalar-valued: $\dot{y} = - y$ (exponential decay)
    - Example: Vector-valued:
      $$
      \begin{aligned}
      \dot{y}_1 &= y_2, \\
      \dot{y}_2 &= -g,
      \end{aligned}
      $$
      where $y_1$ is the position, $y_2$ is the velocity, and $g$ is the acceleration due to gravity. This represents the motion of a body in freefall under constant gravitational acceleration.
- The time to stop the solution $t_f$ and number of steps to take $N$.

The class will have two functions:

- `f(t,y)` to define the ODE $y' = f(t,y)$
- `solve()` to perform N steps and return the solution as a numpy array

After we build this class together, you will run an exercise to make sure you understand how to make it go, and then you're free to get a jump start on your homework solving a pile of ODEs.

# Build the `ODE_solver` class

```python
import numpy as np

class ODE_solver:
    def __init__(self, t0, tf, y0, N):
        self.y0 = y0
        self.t0 = t0
        self.tf = tf
        self.N = N
        self.h = (tf - t0) / float(N)

        # We need to check what size y0 is and store that information
        self.dim = len(y0) #assumes that y0 is a list or array

        # Create a place to store the solution
        self.t_values = [t0 + i*self.h for i in range(N+1)]  # time values
        self.solution = np.zeros((N+1, self.dim))  # solution array
        self.solution[0, :] = y0  # set initial condition

        print("Initialized")  # success

    def f(self,t,y):
        # Define the right-hand side of the ODE here, assuming a scalar t and a numpy array y
        # Example: dy/dt = -2y (for scalar y),
        # this function should return a numpy array of the same shape as y
        f = np.zeros_like(y)
        f[0] = -2 * y[0]  # Example for the first component
        return f

    def solve(self):
        for n in range(self.N):
            t_n = self.t_values[n]
            y_n = self.solution[n, :]
            f_n = self.f(t_n, y_n)
            self.solution[n+1, :] = y_n + self.h * f_n
        return self.solution

# Instantiate the ODE_solver
Nsteps = 32
mySolve = ODE_solver(0, 2, [3], Nsteps)
t_values = mySolve.t_values
solution = mySolve.solve()

# Plot an analytic solution
t_analytic = np.linspace(0, 2, Nsteps+1)
analytic = 3 * np.exp(-2 * np.array(t_analytic))

import matplotlib.pyplot as plt
plt.plot(t_values, solution[:, 0], label='y1(t)')
plt.plot(t_analytic, analytic, label='Analytic y1(t)', linestyle='dashed')

print(f'The error between the numerical and analytic solution is {np.linalg.norm(solution[:,0] - analytic[:])}')
```

```python
import numpy as np

class ODE_solver:
    def __init__(self, t0, tf, y0, N):
        self.y0 = y0
        self.t0 = t0
        self.tf = tf
        self.N = N
        self.h = (tf - t0) / float(N)

        # We need to check what size y0 is and store that information
        self.dim = len(y0) #assumes that y0 is a list or array

        # Create a place to store the solution
        self.t_values = [t0 + i*self.h for i in range(N+1)]  # time values
        self.solution = np.zeros((N+1, self.dim))  # solution array
        self.solution[0, :] = y0  # set initial condition

        print("Initialized")  # success

    def f(self,t,y):
        # Define the right-hand side of the ODE here, assuming a scalar t and a numpy array y
        # Example: dy/dt = -2y (for scalar y),
        # this function should return a numpy array of the same shape as y
        f = np.zeros_like(y)
        f[0] = y[1]  # Example for the first component
        f[1] = -y[0]  # Example for the second component
        return f

    def solve(self):
        for n in range(self.N):
            t_n = self.t_values[n]
            y_n = self.solution[n, :]
            f_n = self.f(t_n, y_n)
            self.solution[n+1, :] = y_n + self.h * f_n
        return self.solution

# Instantiate the ODE_solver
Nsteps = 5000
mySolve = ODE_solver(0, 4.*np.pi, [0,1], Nsteps)
t_values = mySolve.t_values
solution = mySolve.solve()

# # Plot an analytic solution
t_analytic = np.linspace(0, 4.*np.pi, Nsteps+1)
analytic1 = [np.sin(t) for t in t_analytic]
analytic2 = [np.cos(t) for t in t_analytic]

import matplotlib.pyplot as plt
plt.plot(t_values, solution[:, 0], label='y1(t)')
plt.plot(t_values, solution[:, 1], label='y2(t)')
plt.plot(t_analytic, analytic1, label='Analytic y1(t)', linestyle='dashed')
plt.plot(t_analytic, analytic2, label='Analytic y2(t)', linestyle='dashed')

print(f'The error between the numerical and analytic solution is {np.linalg.norm(solution[:,0] - analytic1[:])}')
print(f'The error between the numerical and analytic solution is {np.linalg.norm(solution[:,1] - analytic2[:])}')
```
